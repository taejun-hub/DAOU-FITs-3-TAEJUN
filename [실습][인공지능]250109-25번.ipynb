```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 랜덤 데이터 생성
np.random.seed(42)

# 고객 수
n = 1000

# 고객 특성 생성

data = {
    'customer_id': np.arange(1, n + 1),
    'age': np.random.randint(18, 70, size=n), # 18세에서 70세 사이의 나이
    'tenure': np.random.randint(1, 10, size=n), # 가입기간(1에서 10년 사이)
    'balance': np.random.uniform(0, 2500, size=n), # 잔액(0에서 2500 사이)
    'num_products': np.random.randint(1, 4, size=n), # 사용 중인 제품 수 (1에서 3개 사이)
    'has_credit_card': np.random.choice([0, 1], size=n), # 신용카드 보유 여부
    'is_active_member': np.random.choice([0, 1], size=n), # 활동 회원 여부
    'estimated_salary': np.random.uniform(3000, 15000, size=n), # 예상 연봉
    'churn': np.random.choice([0, 1], size=n, p=[0.8, 0.2]) # 고객 이탈 여부(20% 이탈)
}

# 데이터프레임 생성
df = pd.DataFrame(data)

# CSV파일로 저장
df.to_csv('customer_churn.csv', index=False)

# 데이터 확인
print(df.head())
```
