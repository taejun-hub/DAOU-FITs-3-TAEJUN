```python
# 배열의 요소 추가/삽입
arr = [1, 2, 3]
arr.append(4)
arr.insert(1, 10)
arr_ones = [1] * 5

print(arr)
print(arr_ones)
```

```python
# 배열 요소 삭제
arr = [10, 20, 30, 40, 50]
arr.remove(30)
arr.pop()
arr.pop(0)

print(arr)
```

```python
# 배열 요소 수정
arr = [1, 2, 3, 4, 5]
arr[2] = 99

print(arr)
```

```python
# 배열 합치기
arr1 = [1, 2, 3]
arr2 = [4, 5, 6]
arr3 = arr1 + arr2
arr1.extend(arr2)

print(arr3)
print(arr1)
```
